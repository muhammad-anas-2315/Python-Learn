#### While Loop
    where as while loop: condition based

```python
a = 0
while a<10:
    print(a,'Hy')
    a+=1 # increment is very important otherwise loop infinite
```

```python
a = 100
while a>0:
    print(a,'Hy')
    a-=10 # increment is very important otherwise loop infinite
```

```python
#increment
age = 100
age += 10 #age = age + 10
```

```python
guest = []
count = 0
while count <5:
    name = input('Enter guest name: (press q to exit').lower()
    if name == 'q':
        break
    guest.append(name)
    count+=1
guest
```

```python
guest = []
while True:
    name = input('Enter guest name: (press q to exit').lower()
    if name == 'q':
        break
    guest.append(name)
guest
```

```python
cubes = {}
for a in range(1,11):
    value = a**3
    cubes[a] = value
print(cubes)
```

```python
cubes = []
for a in range(1,11):
    value = a**3
    cubes.append(value)
print(cubes[:3])
print(cubes[3:6])
print(cubes[-4:])
```
