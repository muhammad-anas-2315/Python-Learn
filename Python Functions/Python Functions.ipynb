```python
#import string
def text_analyzer(txt):
    length = len(txt)
    words = len(txt.split())
    sentences = len([s for s in txt.split(".") if s.strip()])
    paragraphs = len([p for p in txt.split("\n") if p.strip()])
    vowels_list = "aeiouAEIOU"
    punc = '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~'
    punctuations = 0
    alphabets = 0
    spaces = 0
    numbers = 0
    lower_case = 0
    upper_case = 0
    vowels = 0
    others = 0
    for char in txt:
        if char.isalpha(): #alphabets += chr.isalpha()
            alphabets += 1
            if char.isupper():
                upper_case += 1
            else:
                lower_case += 1
            if char in vowels_list:
                vowels += 1
        elif char.isspace():
            spaces += 1
        elif char.isnumeric():
            numbers += 1
        elif char in punc: #elif ch in string.punctuation:
            punctuations += 1
        else:
            others += 1
    print(f'''
    Length of txt is: {length}
    Number of words in txt are: {words}
    Number of sentences in txt are: {sentences}
    Number of paragraphs in txt are: {paragraphs}
    Number of vowels in txt are: {vowels}
    Number of alphabets in txt are: {alphabets}
    Number of spaces in txt are: {spaces}
    Numbers in txt are: {numbers}
    Number of lower case in txt are: {lower_case}
    Number of upper case in txt are: {upper_case}
    Number of punctuation in txt are: {punctuations}
    others:  {others} 
    ''')
```

```python
shop = [{'item':'flour', 'price':100, 'inventory':250},
        {'item':'sugar', 'price':180, 'inventory':500},
        {'item':'rice', 'price':300, 'inventory':500},
        {'item':'oil', 'price':500, 'inventory':1000},
        {'item':'ghee', 'price':500, 'inventory':500},
        {'item':'tea', 'price':400, 'inventory':100},
        {'item':'soap', 'price':200, 'inventory':50},
        {'item':'salt', 'price':50, 'inventory':100}]
```

```python
def shopping():
    cart = {}
    while True:
        item_name = input('Enter the item you want to buy or (press "q" to exit)').lower()
        if item_name == 'q':
            print('Thank you for shopping')
            break
        found = False
        for s in shop:
            if s ['item'] == item_name:
                found = True
                if item_name in cart.keys():
                    print('Alreadt added in cart')
                    break
                print(f"{item_name} {s['price']} ka hy")
                quantity = float(input('Kitna Chaye?'))
                if s ['inventory'] >= quantity:
                    cart [item_name] = s['price'] * quantity
                    s ['inventory'] -= quantity
                else:
                    print(f'{item_name} itna nahi hai')
                break
        if not found:
            print(f'{item_name} not available')
    generate_bill(cart)
```

```python
def generate_bill(cart):
    for item, price in cart.items():
        print(f'''
        {item} : {price}
        ''')
    gross_total = total(cart)
    discount = apply_discount(gross_total)
    tax = tax_apply(gross_total)
    payable_amount = gross_total - discount + tax
    print(f'''
    Total Amount: {gross_total}
    Applied Discount: {discount}
    Tax Applied: {tax}
    Amount Payable: {payable_amount}
    ''')
```

```python
def total(cart):
    total_amt = sum(cart.values())
    return total_amt
```

```python
def apply_discount(total_amt):
    if total_amt>= 50000:
        discount = total_amt*.10
    elif total_amt>= 25000:
        discount = total_amt*.05
    elif total_amt>= 10000:
        discount = total_amt*.01
    else:
        discount = total_amt*0
    return discount
```

```python
def tax_apply(total_amt):
    if total_amt>= 50000:
        tax = total_amt*.05
    elif total_amt>= 25000:
        tax = total_amt*.025
    elif total_amt>= 10000:
        tax = total_amt*.001
    else:
        tax = total_amt*0
    return tax
```

```python
shopping()
```

```python
all_accounts = []
import random
```

```python
def open_account():
    print('Welcome to Bank')
    acc_title = input('Enter account title: ')
    cnic = input('Enter Cnic No. ')
    contact = input('Enter Contact No. ')
    initial_deposit = int(input('Initial Deposit: '))
    acc_no = random.randint(10000,10000000)
    account = {'title':acc_title,
              'cnic':cnic,
              'contact':contact,
              'balance':initial_deposit,
              'account_no':acc_no}
    all_accounts.append(account)
    print('Your account opened!')
    print(f'Your account title is {account['title']} & Account no is {account['account_no']}')
```

```python
def cash_deposit(account_no, amount):
    if amount > 0:
        for acc in all_accounts:
            if acc['account_no'] == account_no:
                acc['balance'] += amount
                print('Amount deposited successfully')
                break
        else:
            print('invalid account no!')
    else:
        print('invalid ammount!')
```

```python
def add_even_nums(*nums):
    res = 0
    for n in nums:
        if n%2 ==0:
            res +=n
    return res
```

```python
add_even_nums(1,2,3,4,5,6,7,8,10,50)
```

```python
def add_even_nums(a,b,c,*nums):
    res = 0
    for n in nums:
        if n%2 ==0:
            res +=n
    return res, a,b,c
```

#### Method 1

```python
def make_my_profile(name, subject, last_qualification, experience, **other_info):
    return {'Name':name, 'Subject':subject, 'Qualification':last_qualification,'Experience':experience} | other_info
```

```python
make_my_profile(name='nasir',subject='DataScience',last_qualification='MS',experience=20, city='khi')
```

#### Method 2

```python
def make_my_profile(name, subject, last_qualification, experience, **other_info):
    profile = {}
    profile['Name'] = name
    profile['Subject'] = subject
    profile['Qualification'] = last_qualification
    profile['Experience'] = experience
    return profile, other_info
```

```python
make_my_profile(name='nasir',subject='DataScience',last_qualification='MS',experience=20, city='khi', age =25)
```

#### Method 3 (.update)

```python
def make_my_profile(name, subject, last_qualification, experience, **other_info):
    profile = {}
    profile['Name'] = name
    profile['Subject'] = subject
    profile['Qualification'] = last_qualification
    profile['Experience'] = experience
    profile.update(other_info)
    return profile
```

```python
make_my_profile(name='nasir',subject='DataScience',last_qualification='MS',experience=20, city='khi', age =25)
```
