```python
# count 
print(text.count("e")) # string me kisi word ya character kitni baar aya hai wo count karna
print(text.count("er"))
print(text.count("data")) 
```

```python
# center
#(text ko center me lana (beech me rakhna)
print(text.center(130)) # left & right me spaces add ho gayi
print(text.center(130,"-")) # custom characters bhi add krskte ho
```

```python
#### find (index)
#string me kisi word ya character ki position (index) batana
print(text.find("a")) # python me index 0 se start hoti hy
print(text.find("S")) # index (position) space bi count hoti hy
print(text.find("a",1))
print(text.find("A")) # agr word ya character na mila to -1 execute kryga
```

```python
# index
print(text.index("S")) # string me kisi character ya word ki position (index number) batana
# agr word ya character na mila to error ayaga
```

```python
# encode ( .encode() = text ➜ bytes )
# text ko computer-readable byte format me badal deta hai.
# use?
#File save karne me - # Internet data bhejne me
# Network programming me - # Encryption / processing me
```

```python
print(text.encode()) # (string ko bytes me convert karna)
print(text.encode("utf-8")) # Agar kuch na do to default UTF-8 hota hai.
```

```python
pt = "پاکستان"
print(pt.encode("utf-8")) #ye computer language me text ka format hai.
print(pt.encode())
```

```python
# decode ( .decode() = bytes ➜ text )
text_2 = b'a String is a data type found in computer programming that consists of alphanumeric characters'
print(text_2.decode())
```

```python
# endwith (check karna k string kis word/ character par end ho rahi ha)
print(text.endswith("r"))
print(text.endswith("s"))
print(text.endswith("characters"))
```

```python
# startwith (check karna k string kis word/ character par start ho rahi ha)
print(text.startswith("a"))
print(text.startswith("s"))
```

```python
#expand tabs
user= "anas\tUsman" # string me jo \t (tab) hota hai usko spaces me convert karna 
print(user.expandtabs()) #\t = 8 spaces tak expand hota hai
print(user.expandtabs(12))
```
