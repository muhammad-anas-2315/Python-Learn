#### try: except

```python
try:
    file_name = input('Enter file name with extension')
    with open(file_name) as f:
        print(f.read())
except FileNotFoundError:
    print('The file you are looking for in not in this directory')
```

```python
try:
    file_name = input('Enter file name with extension')
    lst = [1,3,5,7,9]
    print(lst[3])
    c=12/0
    with open(file_name) as f:
        print(f.read())
except FileNotFoundError:
    print('The file you are looking for in not in this directory')
except IndexError:
    print('Please Write Correct Index Number')
except ZeroDivisionError:
    print('0 se Divide?')
```

#### else

```python
try:
    n1 = int(input('Enter a num: '))
    n2 = int(input('Enter a num: '))
    res = n1/n2
    print(res)
except Exception as e:
    print('Exception Caught:', e)
else:
    print(f'Our favourite number is {res*200}')
```

#### finally

```python
try:
    n1 = int(input('Enter a num: '))
    n2 = int(input('Enter a num: '))
    res = n1/n2
    print(res)
except Exception as e:
    print('Exception Caught:', e)
else:
    print(f'Our favourite number is {res*200}')
finally:
    print('ye to lazmi chlyga!!')
```

```python
try:
    f = open('test.txt','r+')
    age = input('Enter age: ')
    f.write(age)
    name = input('Enter name: ')
    f.write(name)
    weight = input('Enter weight: ')
    f.write(weight)
except FileNotFoundError:
    print('The file you are accessing is not found')
finally:
    f.close()
```

#### User Defined Exceptions

```python
try:
    age = int(input('Enter Student age: ')) # 15 < age > 70 error create
    if (age<15) or (age>70):
        raise Exception('Student age less than 15 & greater than 70 is not eligible')
except Exception as e:
    print(e)
```
