```python
arr1d = np.random.random_integers(1,21,20)
arr1d
```

```python
arr1d [~(arr1d >= 10)] # tild sign reverse (not)
```

```python
arr1d[arr1d % 2 == 0]
```

```python
arr = np.array([10,15,20,25,30])
result = np.where(arr > 20) # find indices of elements greater
print(result)
```

```python
arr = np.array([10,15,20,25,30])
result = np.where(arr > 20, 1,0) # replace value based on condition if value true then 1 otherwise 0
print(result)
```

```python
arr1 = np.array([10,15,20,25,30])
arr2 = np.array([100,150,200,250,300])
result= np.where(arr1 > 20, arr1, arr2) # to select elements from arr1 where the condition is true
print(result)
```
