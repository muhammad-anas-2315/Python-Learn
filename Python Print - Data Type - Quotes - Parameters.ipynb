#### Parameters Functions

###### Parameters (end="") Ye decide karta hai ke line ke end par kya ayega
- What is parameter:
    - Parameter → jo function me already defined hota hai
    - Argument → jo value hum dete hain
- \n means move to next line
- \t means tab space

```python
print('anas arshad') # is condition me line change ki wajah by default parameter (end=\n) add hy
print('Data Science & AI')
print('anas arshad',end="") # ab is parameter (end=\n) add krny pr value single line ho jae gi
print('Data Science & AI')
print('anas arshad',end="\t") # ab is parameter (end=\n) or |t add krny pr tab space add ho jae ga
print('Data Science & AI')
print('Python is\nprogramming language') #isko direct text k sath bhi likha jaskta hy
print('Python''\n''easy programming language') #seperate me text k sath bhi likha jaskta hy
```

######  Parameter Seperate (sep=) Ye decide karta hai ke values ke darmiyan kya ayega

```python
print("AnasArshad","Usman","NasirHussain") # Ye 3 Values Hain ink darmian me space ki waja by default (sep=' ') laga hua hy
print("AnasArshad","Usman","NasirHussain",sep='') # ab is parameter (sep='') ko add krny pr space remove

print("AnasArshad","Usman","NasirHussain",sep='\n')
```
