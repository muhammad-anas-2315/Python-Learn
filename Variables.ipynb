      True - class - finally - is - return
      and - continue - for - lambda - try
      as - def - from - nonlocal - while

      async - elif - if - or - yield

#### Conventions to declare a variable
- Key Naming Convention Styles
    - Snake Case:
           All letters are lowercase, with words separated by underscores
    - Camel Case:
          The first letter is lowercase, and each subsequent word starts with a capital letter
    - Pascal Case:
          Every word starts with a capital letter
    - SCREAMING_SNAKE_CASE:
          All uppercase letters, separated by underscores. Usually reserved for constants
    - kebab-case:
          Words are separated by dashes

```python
# snake case (all letter lowercase word seperated with underscore)
father_name = "arshad"
my_first_programming_book = "python" 
```

```python
# camel case (first letter lowercase & eacg subsequent word start with capital letter)
myFirstProgrammingBook = "Python Crash"
```
