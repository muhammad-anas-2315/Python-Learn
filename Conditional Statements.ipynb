#### If / Elif

```python
# 600000 se kam 0 tax
# 600000 se 1200000 5% tax
# 1200000 se 2400000 10% tax
# 2400000 se above 15% tax
salary = int(input("Enter Salary: "))
if salary >= 2400000:
    tax_amount = salary*0.15
    net = salary-tax_amount
elif salary >= 1200000:
    tax_amount = salary*0.10
    net = salary-tax_amount
elif salary >= 600000:
    tax_amount = salary*0.05
    net = salary-tax_amount
else:
    tax_amount = salary*0

print(f"You Salary is: {salary}")
print(f"Your Tax: {tax_amount}")
print(f"Your Net Salary: {net}")
```

```python
math = int(input("Enter your Math Marks: "))
eng = int(input("Enter your English Marks: "))
urdu = int(input("Enter your Urdu Marks: "))
science = int(input("Enter your Science Marks: "))
pst = int(input("Enter your PST Marks: "))

total_obt = math + eng + urdu + science + pst
percentage = total_obt / 500*100
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"
print(f"""
------------- Mark Sheet --------------
Your Total Obtain Marks is: {total_obt}
Your Percentage is        : {percentage}
Your Grade is             : {grade}
""")
```

#### Nested If

```python
election_day = input("Aj Election Day Hai?").lower()
if election_day == "yes":
    cnic = input("Cnic Laye Ho?").lower()
    if cnic == "yes":
        voter_list = input("Apka Naam Voter List Me Hai?").lower()
        if voter_list == "yes":
            print("Welcome!")
        else:
            print("Apka Naam Voter List Me Nahi Hai!")
    else:
        print("Without CNIC Not Allowed!")
else:
    print("Nahi Hai Jao")
```
