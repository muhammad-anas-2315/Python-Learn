```python
class Laptop: #protected class
    def __init__(self, model, ram, storage, processor):
        self._model = model
        self._ram = ram
        self._storage = storage
        self._processor = processor
    def get_model(self):
        return self._model
    def set_model(self, new_model):
        self._model = new_model
    def get_ram(self):
        return self._ram
    def set_ram(self, new_ram):
        if new_ram > 16:
            print(f'{new_ram} gb is not supported')
        else:
            self._ram = new_ram
            print(f'Ram updated to {new_ram} gb')
    def get_storage(self):
        return self._storage
    def set_storage(self, new_storage):
        self._storage = new_storage
    def get_processor(self):
        return self._processor
    def set_processor(self, new_processor):
        self._processor = new_processor
```

#### Inheritance:
    - If class A inherits B, then class B is called parent class & class A is called child class
    - The child class will have access to every thing of parent

```python
# Define a class A with two functions:
class A:
    def feature1(self):
        print('feature 1 is working')
    def feature2(self):
        print('feature 2 is working')
a1 = A()
a1.feature1()
a1.feature2()
```

```python
# Define another class with two functions:
class B():
    def feature3(self):
        print('feature 3 is working')
    def feature4(self):
        print('feature 4 is working')
b1 = B()
b1.feature3()
b1.feature4()
```

```python
class B(A): # just by passing class name of parent class
    def feature3(self):
        print('feature 3 is working')
    def feature4(self):
        print('feature 4 is working')
# class was accessing only class B functions
# now class B inherits class A, it can access all the
# attribute & behaviours of class A
b2 = B()
b2.feature1()
b2.feature2()
b2.feature3()
b2.feature4()
```

#### Multilevel Inheritance

```python
# define a class C that inherits class B directly
# and inherits class A indirectly
# class C will have access to features of class A also
class C(B):
    def feature5(self):
        print('feature 5 is working')
    def feature6(self):
        print('feature 6 is working')

c1 = C()
c1.feature1()
c1.feature2()
c1.feature3()
c1.feature4()
c1.feature5()
c1.feature6()
```

#### Abstraction
    Abstract Class & Methods in Python
    - Python does not support abstraction
    - we will use a module ABC fot abstraction
    - ABC means abstract base classes

```python
from abc import ABC
```

```python
# A normal class and a normal method
class Computer:
    def process(self):
        print('running')
```

```python
# A method that only has declaration but has nothing in it is method
class Computer:
    def process(self):
        pass
# A class having a methods that has no body
```

```python
com1 = Computer()
com1.process()

# there is no error and we are able to create object and call method
# because its not an abstract class and not an abstract method.
```

```python
from abc import ABC, abstractmethod
class Computer(ABC):
    @abstractmethod
    def process(self):
        pass
    @abstractmethod
    def greet(self):
        print('Hello')
        # to make a class abstract
        # it must inherit the ABC class from abc module
        # it must have atleast a abstract method
        # (which is defined using a decorator @abstractmethod)
```

```python
#com1 = Computer()

# we can not create the objects of abstract classes
```

```python
class Laptop(Computer):
    def process(self):
        print('it is running')
    def greet(self):
        print('salam')
    def run(self):
        print('running')
lap1 = Laptop()
```

```python
from abc import ABC, abstractmethod
class Car(ABC):
    @abstractmethod
    def mileage(self):
        pass
class Tesla(Car):
    def mileage(self):
        print('The mileage is 30kmph')
class Suzuki(Car):
    def mileage(self):
        print('The mileage is 25kmph')
class Duster(Car):
    def mileage(self):
        print('The mileage is 24kmph')
class Renault(Car):
    def mileage(self):
        print('The mileage is 27kmph')
# Driver code
t = Tesla()
t.mileage()

r = Renault()
r.mileage()

s = Suzuki()
s.mileage()

d = Duster()
d.mileage()
```

```python
from abc import ABC, abstractmethod
class Polygon(ABC):
    #abstract method
    def sides(self):
        pass
class Triangle(Polygon):
    def sides(self):print('Triangle has 3 sides')

class Pentagon(Polygon):
    def sides(self):print('Pentagon has 5 sides')

class Hexagon(Polygon):
    def sides(self):print('Hexagon has 6 sides')

class Square(Polygon):
    def sides(self):print('Square has 4 sides')
        
# Driver code
t = Triangle()
t.sides()

s = Square()
s.sides()

p = Pentagon()
p.sides()

k = Hexagon()
k.sides()
```
