#### Context manager

```python
with open("my file.txt",'r') as f:
    print(f.read())
```

```python
with open("my file.txt", 'r') as f:
    print(f.readline())
```

```python
with open("my file.txt",'r') as f:
    print(f.readlines())
```

```python
with open("my file.txt",'r') as f:
    for line in f:
        print(line)
```

```python
with open("my file.txt",'r') as f:
    for i,line in enumerate(f):
        print(i,line)
```

```python
with open("my file.txt",'r') as f:
    for line in f:
        if len(line)>100:
            continue
        print(line)
```

```python
with open("my file.txt",'r') as f:
    for i,line in enumerate(f):
        if len(line)>100:
            continue
        print(i,line)
```

```python
with open("my file.txt",'r') as f:
    for i,line in enumerate(f):
        if ('open' in line) or ('closed' in line):
            print(i,line)
```

#### Write Text File

```python
with open('guest.txt','w') as f:
    serial = 0
    while True:
        guest = input('Enter guest name: or q to exit').lower()
        if guest!='q':
            f.write(f'{serial}- {guest} \n')
            serial+=1
        else:
            break
```

```python
with open('guest.txt','r') as f:
    print(f.read())
```

```python
with open('my file.txt','r+') as file: # read & write
    print(file.read())
    file.write('\n This is new line')
```

```python
with open('my file.txt','r+') as file: # read & write
    print(file.read())
    file.write('\n This is new line')
    file.seek(0)
    print(file.read())
```
